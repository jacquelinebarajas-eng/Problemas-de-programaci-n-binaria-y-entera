```python
# Import functions from the docplex module
from docplex.mp.model import Model
import math

m = Model("Problema de los asignación de órdenes a máquina")

# Conjuntos
orden = [(j) for j in range(5)]  # 1,2,3,4,5
maquina = [(i) for i in range(5)]    # 1,2,3,4,5

# Parámetro: c[i][j] = costo procesar la orden i en la máquina j
c = [[16, 4, 9, 5, 6],
    [2, 14, 7, 5, 13],
    [8, 10, 3, 12, 11],
    [3, 7, 6, 10, 5],
    [3, 6, 8, 11, 7],]

# Definir variables de decision
x = m.binary_var_matrix(orden, maquina, name='x')

# Definir las restricciones
m.add_constraints(m.sum(x[i, j] for i in orden) == 1 for j in maquina)
m.add_constraints(m.sum(x[i, j] for j in maquina) == 1 for i in orden)

# Definir la funcion objetivo
m.minimize(m.sum(c[i][j] * x[i, j] for i in orden for j in maquina))

# Resolver e imprimir solucion
solution = m.solve(log_output=True,)
print(solution)

# Imprimir la asignacion
nombre_orden = ['A1','A2','A3','A4','A5']
nombre_maquina = ['M1','M2','M3','M4','M5']

if solution:
   print("Número de órdenes = ", m.objective_value)
   for i in orden:
    for j in maquina:
      if x[i,j].solution_value > 0.5:
                  print("Orden", nombre_orden[i], "-> Maquina",
                  nombre_maquina[j], "(costo", c[i][j], ")")

m.export_as_lp(path=".")
```
